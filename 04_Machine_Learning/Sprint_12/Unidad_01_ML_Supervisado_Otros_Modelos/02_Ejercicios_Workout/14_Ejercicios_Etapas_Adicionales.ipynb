```python
import numpy as np
from catboost import CatBoostClassifier, Pool, cv
from sklearn.metrics import balanced_accuracy_score

train_pool = Pool(X_train, y_train)
test_pool  = Pool(X_test, y_test)

params = {
    "loss_function": "MultiClass",
    "eval_metric": "Accuracy",      # ✅ compatible with MultiClass
    "depth": 5,
    "auto_class_weights": "Balanced",
    "iterations": 500,
    "learning_rate": 0.1,
    "random_seed": 42,
    "verbose": False
}

cv_res = cv(
    pool=train_pool,
    params=params,
    fold_count=5,
    shuffle=True,
    partition_random_seed=42
)

# CV Accuracy (mean of last iteration)
acc_col = [c for c in cv_res.columns if "test-Accuracy" in c and "mean" in c][0]
print("CV Accuracy:", round(float(cv_res[acc_col].iloc[-1]), 4))

# Train final model
cat_clf = CatBoostClassifier(**params)
cat_clf.fit(train_pool)

# Test balanced_accuracy (required)
y_pred = np.array(cat_clf.predict(test_pool)).reshape(-1)
print("TEST balanced_accuracy:", round(balanced_accuracy_score(y_test, y_pred), 4))
```

```python
# El CatBoost ha conseguido una accuracy de CV y de test bastante soluda, con un 77.04 y 77.92 respectivamente. 
# Por lo que demuestra que es estable y que no ha hecho overfitting. Seguramente ha resuelto el problema con "Very good", "Ideal" y "Good", 
# que son las clases minoritarias, y ha conseguido subir su recall.
```
